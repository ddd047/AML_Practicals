```python
# Implementation of Naive Bayes Classification
# Using Car Dataset

# Training dataset
# Brand, Fuel, Transmission, Engine, Category

data = [
    ['Maruti', 'Petrol', 'Manual', 'Small', 'Economy'],
    ['Hyundai', 'Petrol', 'Manual', 'Small', 'Economy'],
    ['Tata', 'Diesel', 'Manual', 'Medium', 'Economy'],
    ['Honda', 'Petrol', 'Automatic', 'Medium', 'Premium'],
    ['Toyota', 'Diesel', 'Automatic', 'Large', 'Premium'],
    ['BMW', 'Diesel', 'Automatic', 'Large', 'Premium'],
    ['Audi', 'Petrol', 'Automatic', 'Large', 'Premium'],
    ['Maruti', 'Petrol', 'Manual', 'Small', 'Economy'],
    ['Hyundai', 'Diesel', 'Manual', 'Medium', 'Economy'],
    ['Mercedes', 'Diesel', 'Automatic', 'Large', 'Premium'],
    ['Tata', 'Petrol', 'Manual', 'Small', 'Economy'],
    ['Honda', 'Diesel', 'Automatic', 'Medium', 'Premium']
]

# Feature names
features = ['Brand', 'Fuel', 'Transmission', 'Engine']

# Naive Bayes function
def naive_bayes(data, test_data):

    total = len(data)

    # Find unique classes
    classes = set(row[-1] for row in data)

    probabilities = {}

    for cls in classes:

        # Calculate prior probability
        class_data = [row for row in data if row[-1] == cls]

        class_count = len(class_data)
        probability = class_count / total

        # Calculate likelihood using Laplace smoothing
        for i, value in enumerate(test_data):

            count = sum(
                1 for row in class_data if row[i] == value
            )

            unique_values = set(row[i] for row in data)

            probability *= (count + 1) / (
                class_count + len(unique_values)
            )

        probabilities[cls] = probability

    # Find class with highest probability
    prediction = max(probabilities, key=probabilities.get)

    return prediction, probabilities


# Test car data
test_data = ['Honda', 'Petrol', 'Automatic', 'Medium']

# Display results
print("Naive Bayes Car Classification")
print("--------------------------------")

print("Test Car Details:")

for feature, value in zip(features, test_data):
    print(feature, ":", value)

prediction, probabilities = naive_bayes(data, test_data)

print("\nClass Probabilities:")

for cls, prob in probabilities.items():
    print(cls, ":", round(prob, 6))

print("\nPredicted Car Category:", prediction)

print("\nClassification completed successfully!")
```
