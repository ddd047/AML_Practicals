```python
import numpy as np

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# XOR and XNOR expected outputs
xor_y = np.array([[0], [1], [1], [0]])
xnor_y = np.array([[1], [0], [0], [1]])

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Train Multilayer Perceptron
def train_mlp(X, y):

    np.random.seed(42)

    # Initialize weights and biases
    W1 = np.random.randn(2, 4)
    b1 = np.zeros((1, 4))

    W2 = np.random.randn(4, 1)
    b2 = np.zeros((1, 1))

    learning_rate = 1.0

    # Training process
    for epoch in range(20000):

        # Forward propagation
        H = sigmoid(np.dot(X, W1) + b1)
        Y = sigmoid(np.dot(H, W2) + b2)

        # Backpropagation
        error = Y - y

        dW2 = np.dot(H.T, error) / len(X)
        db2 = np.mean(error, axis=0, keepdims=True)

        dH = np.dot(error, W2.T) * H * (1 - H)

        dW1 = np.dot(X.T, dH) / len(X)
        db1 = np.mean(dH, axis=0, keepdims=True)

        # Update weights and biases
        W2 -= learning_rate * dW2
        b2 -= learning_rate * db2

        W1 -= learning_rate * dW1
        b1 -= learning_rate * db1

    return W1, b1, W2, b2


# Test the trained MLP
def display_results(name, y):

    W1, b1, W2, b2 = train_mlp(X, y)

    H = sigmoid(np.dot(X, W1) + b1)
    predictions = sigmoid(np.dot(H, W2) + b2)

    print("\n", name, "GATE")
    print("-------------------------")
    print("A\tB\tOutput")

    for i in range(len(X)):
        output = 1 if predictions[i][0] >= 0.5 else 0

        print(X[i][0], "\t", X[i][1], "\t", output)


# Display XOR results
display_results("XOR", xor_y)

# Display XNOR results
display_results("XNOR", xnor_y)

print("\nMLP Training Completed Successfully!")
```
