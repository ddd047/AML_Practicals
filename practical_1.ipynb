```python
# Implementation of AND, OR, NAND and NOR gates using Perceptron

# Step activation function
def activation(z):
    return 1 if z >= 0 else 0


# Perceptron training function
def train_perceptron(inputs, targets, learning_rate=0.1, epochs=100):
    weights = [0.0, 0.0]
    bias = 0.0

    for epoch in range(epochs):
        for x, target in zip(inputs, targets):

            # Calculate weighted sum
            z = weights[0] * x[0] + weights[1] * x[1] + bias

            # Predict output
            prediction = activation(z)

            # Calculate error
            error = target - prediction

            # Update weights and bias
            weights[0] += learning_rate * error * x[0]
            weights[1] += learning_rate * error * x[1]
            bias += learning_rate * error

    return weights, bias


# Function to test a trained perceptron
def test_perceptron(inputs, weights, bias):
    results = []

    for x in inputs:
        z = weights[0] * x[0] + weights[1] * x[1] + bias
        results.append(activation(z))

    return results


# Input combinations
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

# Truth tables for logic gates
gates = {
    "AND":  [0, 0, 0, 1],
    "OR":   [0, 1, 1, 1],
    "NAND": [1, 1, 1, 0],
    "NOR":  [1, 0, 0, 0]
}

# Train and test each logic gate
for gate, targets in gates.items():

    weights, bias = train_perceptron(inputs, targets)

    outputs = test_perceptron(inputs, weights, bias)

    print("\n", gate, "GATE")
    print("---------------------")
    print("Input1  Input2  Output")

    for x, output in zip(inputs, outputs):
        print(f"  {x[0]}       {x[1]}       {output}")
```
