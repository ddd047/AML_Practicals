```python
# Implementation of Artificial Neural Network (ANN)

import numpy as np

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


# Input data (XOR problem)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Expected output
Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Initialize random weights
np.random.seed(42)

W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))

W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

# Learning rate
learning_rate = 0.5

# Training the neural network
for epoch in range(10000):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate error
    error = predicted_output - Y

    # Backpropagation
    d_output = error * predicted_output * (1 - predicted_output)

    dW2 = np.dot(hidden_output.T, d_output) / len(X)
    db2 = np.mean(d_output, axis=0, keepdims=True)

    d_hidden = np.dot(d_output, W2.T) * hidden_output * (1 - hidden_output)

    dW1 = np.dot(X.T, d_hidden) / len(X)
    db1 = np.mean(d_hidden, axis=0, keepdims=True)

    # Update weights and biases
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1


# Display final results
print("Artificial Neural Network (ANN)")
print("--------------------------------")

print("Input\t\tExpected\tPredicted")

for i in range(len(X)):
    print(
        X[i],
        "\t",
        Y[i][0],
        "\t\t",
        round(predicted_output[i][0], 4)
    )

print("\nTraining completed successfully!")
```
