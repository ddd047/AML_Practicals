```python
import numpy as np

# Define objective function
def objective_function(x):
    return np.sum(x ** 2)

# PSO parameters
num_particles = 30
dimensions = 2
iterations = 100

w = 0.7       # Inertia weight
c1 = 1.5      # Cognitive coefficient
c2 = 1.5      # Social coefficient

# Initialize particles and velocities
np.random.seed(42)

positions = np.random.uniform(
    -10, 10, (num_particles, dimensions)
)

velocities = np.random.uniform(
    -1, 1, (num_particles, dimensions)
)

# Initialize personal best
pbest_positions = positions.copy()
pbest_values = np.array([
    objective_function(p) for p in positions
])

# Initialize global best
best_index = np.argmin(pbest_values)

gbest_position = pbest_positions[best_index].copy()
gbest_value = pbest_values[best_index]

# PSO main loop
for iteration in range(iterations):

    for i in range(num_particles):

        r1 = np.random.rand(dimensions)
        r2 = np.random.rand(dimensions)

        # Update velocity
        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pbest_positions[i] - positions[i])
            + c2 * r2 * (gbest_position - positions[i])
        )

        # Update position
        positions[i] += velocities[i]

        # Keep particles within search limits
        positions[i] = np.clip(positions[i], -10, 10)

        # Calculate fitness
        fitness = objective_function(positions[i])

        # Update personal best
        if fitness < pbest_values[i]:
            pbest_values[i] = fitness
            pbest_positions[i] = positions[i].copy()

    # Update global best
    best_index = np.argmin(pbest_values)

    if pbest_values[best_index] < gbest_value:
        gbest_value = pbest_values[best_index]
        gbest_position = pbest_positions[best_index].copy()

# Display results
print("Particle Swarm Optimization")
print("---------------------------")

print("Best Position:", gbest_position)
print("Minimum Fitness Value:", gbest_value)

print("Iterations:", iterations)
print("\nOptimization Completed Successfully!")
```
