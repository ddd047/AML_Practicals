```python
# Implementation of basic Recurrent Neural Network (RNN)

import numpy as np

# Define input sequence
inputs = [1, 2, 3]

# Initialize weights and biases
Wxh = np.array([[0.5, -0.3]])       # Input to hidden
Whh = np.array([[0.1, 0.2],
                [-0.1, 0.1]])      # Hidden to hidden
Why = np.array([[1.0],
                [0.5]])            # Hidden to output

bh = np.array([[0.0, 0.0]])         # Hidden bias
by = np.array([[0.0]])              # Output bias

# Initialize hidden state
h = np.zeros((1, 2))

print("Recurrent Neural Network (RNN)")
print("--------------------------------")

# Process each input in the sequence
for t, x in enumerate(inputs):

    # Convert input into array
    x = np.array([[x]])

    # Calculate new hidden state
    h = np.tanh(np.dot(x, Wxh) + np.dot(h, Whh) + bh)

    # Calculate output
    y = np.dot(h, Why) + by

    # Display results
    print("\nTime Step:", t + 1)
    print("Input:", x[0][0])
    print("Hidden State:", h)
    print("Output:", y[0][0])

print("\nRNN execution completed successfully!")
```
